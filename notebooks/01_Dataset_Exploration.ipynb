```python
# Prepare the sentiment target variable

from sklearn.preprocessing import LabelEncoder

# Create the encoder
label_encoder = LabelEncoder()

# Convert sentiment labels into numerical values
y = label_encoder.fit_transform(df_clean["Sentiment"])

# Display the mapping
print("SENTIMENT LABEL ENCODING")
print("=" * 50)

for index, label in enumerate(label_encoder.classes_):
    print(label, "=", index)

# Display target information
print("\nTarget variable information")
print("Number of records:", len(y))
print("Number of classes:", len(label_encoder.classes_))
print("Target shape:", y.shape)

print("\nEncoded target distribution:")

import pandas as pd

display(
    pd.Series(y)
    .value_counts()
    .sort_index()
    .rename("Number of Records")
    .to_frame()
    .rename_axis("Encoded Class")
)
```

```python
# Select features for sentiment classification

selected_features = [
    "Text",
    "Title",
    "Location_Type",
    "Province",
    "Location_Name",
    "User_Country",
    "Review_Length",
    "Title_Length"
]

X = df_clean[selected_features].copy()

print("FEATURE SELECTION SUMMARY")
print("=" * 55)

print("Number of input records:", X.shape[0])
print("Number of selected features:", X.shape[1])

print("\nSelected feature names:")
for feature in selected_features:
    print("-", feature)

print("\nFeature data types:")
print(X.dtypes)

print("\nFirst five records:")
display(X.head())

print("\nTarget shape:", y.shape)
print("Input shape:", X.shape)
```

```python
# Create a leakage-safe training and testing split

from sklearn.model_selection import StratifiedGroupKFold
import numpy as np
import pandas as pd

# Configure the grouped split
splitter = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Generate the first 80/20 split
train_indices, test_indices = next(
    splitter.split(X, y, groups=X["Text"])
)

# Create training and testing datasets
X_train = X.iloc[train_indices].copy()
X_test = X.iloc[test_indices].copy()

y_train = y[train_indices]
y_test = y[test_indices]

# Check for duplicate review text leakage
train_texts = set(X_train["Text"])
test_texts = set(X_test["Text"])

overlap = train_texts.intersection(test_texts)

print("TRAIN-TEST SPLIT SUMMARY")
print("=" * 55)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining percentage:",
      round(len(X_train) / len(X) * 100, 2), "%")

print("Testing percentage:",
      round(len(X_test) / len(X) * 100, 2), "%")

print("\nDUPLICATE TEXT LEAKAGE CHECK")
print("=" * 55)

print("Identical review texts appearing in both sets:",
      len(overlap))

print("\nTRAINING SENTIMENT DISTRIBUTION")
print("=" * 55)

display(
    pd.Series(
        label_encoder.inverse_transform(y_train)
    ).value_counts().to_frame("Records")
)

print("\nTESTING SENTIMENT DISTRIBUTION")
print("=" * 55)

display(
    pd.Series(
        label_encoder.inverse_transform(y_test)
    ).value_counts().to_frame("Records")
)
```

```python
# Import preprocessing tools

from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Text preprocessing
text_transformer = TfidfVectorizer(
    max_features=3000,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

title_transformer = TfidfVectorizer(
    max_features=1000,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

# Numerical preprocessing
numeric_transformer = StandardScaler()

# Categorical preprocessing
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)

# Combine all transformations
preprocessor = ColumnTransformer(
    transformers=[
        ("review_text", text_transformer, "Text"),
        ("title_text", title_transformer, "Title"),
        (
            "categorical",
            categorical_transformer,
            [
                "Location_Type",
                "Province",
                "Location_Name",
                "User_Country"
            ]
        ),
        (
            "numerical",
            numeric_transformer,
            [
                "Review_Length",
                "Title_Length"
            ]
        )
    ]
)

print("PREPROCESSING PIPELINE CREATED")
print("=" * 55)

print(preprocessor)

print("\nText features: TF-IDF")
print("Categorical features: One-Hot Encoding")
print("Numerical features: Standard Scaling")
```

```python
# Fit preprocessing using training data only

print("FITTING PREPROCESSING PIPELINE")
print("=" * 55)

X_train_processed = preprocessor.fit_transform(X_train)

print("Training data transformed successfully!")

# Transform testing data without fitting again
X_test_processed = preprocessor.transform(X_test)

print("Testing data transformed successfully!")

# Display transformation information
print("\nTRANSFORMED DATA SUMMARY")
print("=" * 55)

print("Original training shape:", X_train.shape)
print("Original testing shape:", X_test.shape)

print("\nProcessed training shape:",
      X_train_processed.shape)

print("Processed testing shape:",
      X_test_processed.shape)

print("\nTraining matrix type:",
      type(X_train_processed))

print("Testing matrix type:",
      type(X_test_processed))

print("\nNumber of transformed features:",
      X_train_processed.shape[1])

print("\nNumber of non-zero training values:",
      X_train_processed.nnz)

print("\nPreprocessing completed successfully!")
```

```python
# Import Decision Tree classifier and evaluation tools

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
import time

# Create the Decision Tree model
decision_tree = DecisionTreeClassifier(
    random_state=42,
    class_weight="balanced",
    max_depth=30,
    min_samples_split=10,
    min_samples_leaf=5
)

# Train the model
print("TRAINING DECISION TREE")
print("=" * 55)

start_time = time.time()

decision_tree.fit(X_train_processed, y_train)

training_time = time.time() - start_time

# Generate predictions
start_time = time.time()

dt_predictions = decision_tree.predict(X_test_processed)

prediction_time = time.time() - start_time

# Evaluate performance
print("\nDECISION TREE RESULTS")
print("=" * 55)

print("Accuracy:",
      round(accuracy_score(y_test, dt_predictions), 4))

print("Precision (Macro):",
      round(precision_score(
          y_test, dt_predictions,
          average="macro", zero_division=0
      ), 4))

print("Recall (Macro):",
      round(recall_score(
          y_test, dt_predictions,
          average="macro", zero_division=0
      ), 4))

print("F1 Score (Macro):",
      round(f1_score(
          y_test, dt_predictions,
          average="macro", zero_division=0
      ), 4))

print("\nTraining time:", round(training_time, 2), "seconds")
print("Prediction time:", round(prediction_time, 4), "seconds")

print("\nDETAILED CLASSIFICATION REPORT")
print("=" * 55)

print(
    classification_report(
        y_test,
        dt_predictions,
        labels=[0, 1, 2],
        target_names=label_encoder.classes_,
        zero_division=0
    )
)
```

```python
from xgboost import XGBClassifier
from sklearn.utils.class_weight import compute_sample_weight
import time

# Calculate balanced sample weights
sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train
)

# Initialize XGBoost
xgb_model = XGBClassifier(
    objective="multi:softprob",
    num_class=3,
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    eval_metric="mlogloss",
    tree_method="hist",
    n_jobs=-1,
    random_state=42
)

print("TRAINING XGBOOST")
print("=" * 55)

# Training
start_time = time.time()

xgb_model.fit(
    X_train_processed,
    y_train,
    sample_weight=sample_weights
)

xgb_training_time = time.time() - start_time

# Prediction
start_time = time.time()

xgb_predictions = xgb_model.predict(X_test_processed)

xgb_prediction_time = time.time() - start_time

# Evaluation
print("\nXGBOOST RESULTS")
print("=" * 55)

print("Accuracy:", round(
    accuracy_score(y_test, xgb_predictions), 4
))

print("Precision (Macro):", round(
    precision_score(y_test, xgb_predictions,
                    average="macro", zero_division=0), 4
))

print("Recall (Macro):", round(
    recall_score(y_test, xgb_predictions,
                 average="macro", zero_division=0), 4
))

print("F1 Score (Macro):", round(
    f1_score(y_test, xgb_predictions,
             average="macro", zero_division=0), 4
))

print("\nTraining time:", round(xgb_training_time, 2), "seconds")
print("Prediction time:", round(xgb_prediction_time, 4), "seconds")

print("\nDETAILED CLASSIFICATION REPORT")
print("=" * 55)

print(classification_report(
    y_test,
    xgb_predictions,
    labels=[0, 1, 2],
    target_names=label_encoder.classes_,
    zero_division=0
))
```

```python
from sklearn.ensemble import StackingClassifier, ExtraTreesClassifier
from sklearn.linear_model import LogisticRegression
import time

# Define base models
base_models = [
    (
        "random_forest",
        RandomForestClassifier(
            n_estimators=100,
            max_features="sqrt",
            class_weight="balanced",
            n_jobs=1,
            random_state=42
        )
    ),
    (
        "extra_trees",
        ExtraTreesClassifier(
            n_estimators=100,
            max_features="sqrt",
            class_weight="balanced",
            n_jobs=1,
            random_state=42
        )
    ),
    (
        "logistic_regression",
        LogisticRegression(
            max_iter=1000,
            class_weight="balanced",
            random_state=42
        )
    )
]

# Define stacking model
stacking_model = StackingClassifier(
    estimators=base_models,
    final_estimator=LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ),
    cv=3,
    stack_method="predict_proba",
    n_jobs=-1
)

print("TRAINING STACKING ENSEMBLE")
print("=" * 55)

# Training
start_time = time.time()

stacking_model.fit(X_train_processed, y_train)

stacking_training_time = time.time() - start_time

# Prediction
start_time = time.time()

stacking_predictions = stacking_model.predict(X_test_processed)

stacking_prediction_time = time.time() - start_time

# Evaluation
print("\nSTACKING ENSEMBLE RESULTS")
print("=" * 55)

print("Accuracy:", round(
    accuracy_score(y_test, stacking_predictions), 4
))

print("Precision (Macro):", round(
    precision_score(y_test, stacking_predictions,
                    average="macro", zero_division=0), 4
))

print("Recall (Macro):", round(
    recall_score(y_test, stacking_predictions,
                 average="macro", zero_division=0), 4
))

print("F1 Score (Macro):", round(
    f1_score(y_test, stacking_predictions,
             average="macro", zero_division=0), 4
))

print("\nTraining time:", round(stacking_training_time, 2), "seconds")
print("Prediction time:", round(stacking_prediction_time, 4), "seconds")

print("\nDETAILED CLASSIFICATION REPORT")
print("=" * 55)

print(classification_report(
    y_test,
    stacking_predictions,
    labels=[0, 1, 2],
    target_names=label_encoder.classes_,
    zero_division=0
))
```

```python

import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Collect all trained models and their predictions
model_results = [
    ("Decision Tree", dt_predictions, training_time, prediction_time),
    ("Bagging", bagging_predictions, bagging_training_time, bagging_prediction_time),
    ("Random Forest", rf_predictions, rf_training_time, rf_prediction_time),
    ("AdaBoost", ada_predictions, ada_training_time, ada_prediction_time),
    ("Gradient Boosting", gb_predictions, gb_training_time, gb_prediction_time),
    ("XGBoost", xgb_predictions, xgb_training_time, xgb_prediction_time),
    ("Stacking", stacking_predictions, stacking_training_time, stacking_prediction_time)
]

# Calculate evaluation metrics
results = []

for name, predictions, train_time, predict_time in model_results:
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Macro Precision": precision_score(
            y_test, predictions, average="macro", zero_division=0
        ),
        "Macro Recall": recall_score(
            y_test, predictions, average="macro", zero_division=0
        ),
        "Macro F1": f1_score(
            y_test, predictions, average="macro", zero_division=0
        ),
        "Weighted F1": f1_score(
            y_test, predictions, average="weighted", zero_division=0
        ),
        "Training Time (s)": train_time,
        "Prediction Time (s)": predict_time
    })

# Create DataFrame
results_df = pd.DataFrame(results)

print("ENSEMBLELAB - MODEL PERFORMANCE COMPARISON")
print("=" * 90)

display(
    results_df.style.format({
        "Accuracy": "{:.4f}",
        "Macro Precision": "{:.4f}",
        "Macro Recall": "{:.4f}",
        "Macro F1": "{:.4f}",
        "Weighted F1": "{:.4f}",
        "Training Time (s)": "{:.2f}",
        "Prediction Time (s)": "{:.4f}"
    })
)
```
